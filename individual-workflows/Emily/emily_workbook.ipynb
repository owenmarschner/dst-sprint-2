Parts to explore:

It is recommended to use lower learning rate then what it was trained on originally to avoid large changes to model wieghts.

About Resnet-18: ResNet-18 is a relatively lightweight 18-layer CNN that uses skip connections to improve training and image feature extraction. It is widely used for image classification, feature extraction, and transfer learning.
